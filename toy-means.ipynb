#### Functions

```python
def sample_normal(seed: int, mean: float = 0, std: float = 1, size: int = 100000) -> Series:
    rng = np.random.default_rng(seed=seed)
    samples = rng.normal(loc=mean, scale=std, size=size)
    return Series(samples, name=rf'$\mathcal{{N}}({mean},{std})$')
```

#### Distributions

```python
SEED = 3
```

```python
P1 = sample_normal(seed=SEED, mean=8, std=1)
P2 = sample_normal(seed=SEED, mean=8, std=2.5)
```

#### Plot

```python
plot_data = pd.concat([P1, P2], axis='columns')
ax = plot_data.plot.kde(style=['--', ':'], xlim=(0, 16), ylim=(0, .42), color='black')
_ = ax.set_ylabel(None)
fig = ax.get_figure()
fig.savefig('data/toy/toy-means.pdf')
```
