# Initialise centroids

#### Libraries

```python
from importlib import import_module

from src import init
```

#### Parameters

```python
DATASET_NAME = 'adult'           # ['adult', 'compas', 'crime', 'motor']
N_CLUSTERS_LIST = range(6, 9)
N_RANDOM_STATES = 5
INIT_METHOD = 'kmeans_plusplus'
```

#### Load dataset

```python
dataset_module = import_module(f"src.datasets.{DATASET_NAME}")
_, X, s, _ = dataset_module.load()
```

#### Initialise centroids

```python
init_method = getattr(import_module("src.init"), INIT_METHOD)
for n_clusters in N_CLUSTERS_LIST:
    for random_state in range(N_RANDOM_STATES):
        centroids = init_method(
            X=X.copy(), n_clusters=n_clusters, random_state=random_state
            )
        init.export(
            centroids=centroids, dataset_name=DATASET_NAME,
            init_method=INIT_METHOD, random_state=random_state
            )
```
